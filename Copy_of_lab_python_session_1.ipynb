````python
from IPython.display import HTML, display

class Progress:
    xp = 0
    exercises_done = []

    def add(self, points, name):
        self.xp += points
        self.exercises_done.append(name)
        self.show()

    def show(self):
        percent = (self.xp / 100) * 100
        bar = "█" * int(percent/10) + "░" * (10 - int(percent/10))
        print(f"\n📊 Progress: [{bar}] {self.xp}/100 XP ({percent:.0f}%)")

        if self.xp >= 100:
            display(HTML("""
            <div style='background: gold; padding: 30px; border-radius: 15px;
                        text-align: center; color: #333;'>
                <h1>🏆 100 XP! SESSION COMPLETE! 🏆</h1>
                <p style='font-size: 24px;'>⭐⭐⭐⭐⭐</p>
            </div>
            """))

progress = Progress()
print("✅ Setup complete! Let's code!")
````

# 📖 LESSON: The print() Function

The `print()` function displays messages.

## Syntax:
```python

```

## Rules:
- ✅ Use parentheses ()
- ✅ Use quotes "" for text
- ✅ Numbers don't need quotes

## Examples:
```python

print("I am", 25)       # Multiple things
```

# 📖 LESSON: Variables

Variables store information!

## Syntax:
```python

```

## Examples:
```python

```

## Rules:
- Use `=` to assign values
- Variable names: letters, numbers, underscore
- Can't start with a number

# 📖 LESSON: F-Strings

F-strings combine text and variables!

## Syntax:
```python
f"Text {variable} more text"
```

Note the `f` before the quote!

## Example:
```python
name = "Alice"
age = 25
print(f"Hi, I'm {name} and I'm {age} years old")
# Output: Hi, I'm Alice and I'm 25 years old
```

---

🎯 Try it!

````python
name = "Python"
version = 3.11

print(f"I'm learning {name} version {version}!")
print(f"{name} is awesome!")

print("\n✨ F-strings make combining text and variables easy!")
````

# 📖 LESSON: Getting User Input

The `input()` function asks users for information!

## Syntax:
```python
variable = input("Your question: ")
```

## Example:
```python

print(f"Hello, {name}!")
```

When you run this:
1. Python asks the question
2. User types answer
3. Answer is stored in variable
